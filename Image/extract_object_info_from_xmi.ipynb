```python
import os  # Import the os module to interact with the operating system
from glob import glob  # Import glob to find all pathnames matching a specified pattern
import pandas as pd  # Import pandas for data manipulation and analysis
from xml.etree import ElementTree as et  # Import ElementTree for parsing XML data
from functools import reduce  # Import reduce to apply a function cumulatively to items in an iterable
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
# step-1: get path of each xml file
xmlfiles = glob('./data_images/*.xml')
xmlfiles
```

```python
def extract_text(filename):
    tree = et.parse(filename)
    root = tree.getroot()

    # Extract filename, width, and height
    image_name = root.find('filename').text if root.find('filename') is not None else None
    width = root.find('size/width').text if root.find('size/width') is not None else None
    height = root.find('size/height').text if root.find('size/height') is not None else None
    
    parser = []
    for obj in root.findall('object'):
        name = obj.find('name').text if obj.find('name') is not None else None
        bndbox = obj.find('bndbox')
        if bndbox is not None:
            xmin = bndbox.find('xmin').text if bndbox.find('xmin') is not None else None
            xmax = bndbox.find('xmax').text if bndbox.find('xmax') is not None else None
            ymin = bndbox.find('ymin').text if bndbox.find('ymin') is not None else None
            ymax = bndbox.find('ymax').text if bndbox.find('ymax') is not None else None
        else:
            xmin = xmax = ymin = ymax = None
        
        parser.append([image_name, width, height, name, xmin, xmax, ymin, ymax])
        
    return parser
```

```python
parser_all = list(map(extract_text,xmlfiles))
parser_all
```

```python
data = reduce(lambda x, y : x+y,parser_all)
data
```

```python
df = pd.DataFrame(data,columns = ['filename','width','height','name','xmin','xmax','ymin','ymax'])
df
```

# conversion




















```python
# type conversion
cols = ['width','height','xmin','xmax','ymin','ymax']
df[cols] = df[cols].astype(int)
df.info()
```

```python
# center x, center y
df['center_x'] = ((df['xmax']+df['xmin'])/2)/df['width']
df['center_y'] = ((df['ymax']+df['ymin'])/2)/df['height']
# w 
df['w'] = (df['xmax']-df['xmin'])/df['width']
# h 
df['h'] = (df['ymax']-df['ymin'])/df['height']
```

```python
df.head()
```

```python
# label encoding4
def label_encoding(x):
    labels = {'person':0, 'car':1, 'chair':2, 'bottle':3, 'bird':4,
              'sofa':5, 'cycle':6, 'horse':7, 'bus':8}
    try:
        return labels[x]
    except KeyError:
        # Handle cases where 'x' is not found in the labels dictionary
        return -1  # Or any other appropriate value or action
df['id'] = df['name'].apply(label_encoding)
df
 
```

### Save Image and Labels in text

```python
import os
from shutil import move
```

```python

cols = ['filename','id','center_x','center_y', 'w', 'h']
groupby_obj_train = train_df[cols].groupby('filename')
groupby_obj_test = test_df[cols].groupby('filename')
```

```python
import os
import shutil
import glob
import pandas as pd

# Function to save data (both images and labels)
def save_data(filename, folder_path, group_obj):
    # Normalize the filename (remove spaces, convert to lowercase, etc.)
    normalized_filename = os.path.splitext(filename)[0].strip().lower()
    
    # Find matching image files (handles multiple extensions like .jpg, .jpeg, .png, etc.)
    matches = glob.glob(f"data_images/{normalized_filename}.*")  # Matches any extension
    
    if matches:
        src_path = matches[0]  # Pick the first matching file
        dst_path = os.path.join(folder_path, os.path.basename(src_path))
        
        # Check if the source file exists before moving
        if os.path.exists(src_path):
            print(f"Moving {src_path} -> {dst_path}")  # Debugging statement
            shutil.move(src_path, dst_path)
        else:
            print(f"Source file does not exist: {src_path}")
    else:
        print(f"No matching image file found for: {filename}")
    
    # Save the labels as a text file if the filename exists in the group object
    if filename in group_obj.groups:
        text_filename = os.path.join(folder_path, os.path.splitext(filename)[0] + '.txt')
        group_data = group_obj.get_group(filename).set_index('filename')
        group_data.to_csv(text_filename, sep=' ', index=False, header=False)
    else:
        print(f"Filename not found in group object: {filename}")

# Ensure the directories exist
train_folder = 'data_images/train'
test_folder = 'data_images/test'
os.makedirs(train_folder, exist_ok=True)
os.makedirs(test_folder, exist_ok=True)

# Apply the function to training filenames
filename_series_train = pd.Series(groupby_obj_train.groups.keys())
filename_series_train.apply(lambda filename: save_data(filename, train_folder, groupby_obj_train))

# Apply the function to test filenames
filename_series_test = pd.Series(groupby_obj_test.groups.keys())
filename_series_test.apply(lambda filename: save_data(filename, test_folder, groupby_obj_test))
```
