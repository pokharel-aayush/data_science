```python
sns.histplot(data=df,x='Debt_to_Income',bins=30,kde=True)
```

```python
#-3 and 3 standard deviation outside are considered outliers arefound through mean(sum/n)
#z=(x−μ)/σ
UL=df['Debt_to_Income'].mean()+3*df['Debt_to_Income'].std()
UL
```

```python
LL=df['Debt_to_Income'].mean()-3*df['Debt_to_Income'].std()
LL
```

```python
df[(df['Debt_to_Income'] > UL) | (df['Debt_to_Income'] < LL)]
```

```python
#to handel outliers trimming(cons= data can be lossed), gapping(for more data ie we have more datas)
df1=df[(df['Debt_to_Income'] > LL) | (df['Debt_to_Income'] < UL)]
df1
```
